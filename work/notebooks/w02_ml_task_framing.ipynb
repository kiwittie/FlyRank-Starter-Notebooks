## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would be too simple because user engagement can depend on several factors at the same time. For example, a rule could say that content with more clicks should always be shown first, but this may not work for every user or type of content. ML can learn patterns from multiple features and identify combinations that are difficult to describe with simple if-statements. The model output could support the decision of which content to show or prioritize for a user.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

reason = (
    "User engagement can depend on multiple factors, so a simple "
    "if/else rule may not capture all the patterns."
)

print(reason)
```
